```python
# Write Python program to plot time series with trends of sales.

import matplotlib.pyplot as plt
import datetime

# Sample time series data: Monthly sales
months = [
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
]
sales = [220, 245, 260, 300, 280, 310, 330, 360, 390, 420, 450, 480]

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(months, sales, marker='o', linestyle='-', color='teal')

# Adding titles and labels
plt.title("Monthly Sales Trend - Time Series")
plt.xlabel("Month")
plt.ylabel("Sales (in units)")
plt.grid(True)

# Show the plot
plt.tight_layout()
plt.show()
```
