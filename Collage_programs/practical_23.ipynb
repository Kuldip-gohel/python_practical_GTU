```python
# Write a python program to do line plotting, scatter plotting, Bar
# chart, box plot, histogram and pie chart.

import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 17, 19]

# -------- Line Plot --------
plt.figure(figsize=(6, 4))
plt.plot(x, y, marker='o', linestyle='-', color='blue')
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()

# -------- Scatter Plot --------
plt.figure(figsize=(6, 4))
plt.scatter(x, y, color='red', marker='x')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()

# -------- Bar Chart --------
categories = ['A', 'B', 'C', 'D', 'E']
values = [5, 7, 3, 8, 6]

plt.figure(figsize=(6, 4))
plt.bar(categories, values, color='green')
plt.title("Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()

# -------- Box Plot --------
data = [10, 12, 13, 13, 14, 15, 16, 18, 19, 20, 21]

plt.figure(figsize=(6, 4))
plt.boxplot(data)
plt.title("Box Plot")
plt.ylabel("Values")
plt.show()

# -------- Histogram --------
import random
data_hist = [random.randint(1, 100) for _ in range(50)]

plt.figure(figsize=(6, 4))
plt.hist(data_hist, bins=10, color='purple', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value Range")
plt.ylabel("Frequency")
plt.show()

# -------- Pie Chart --------
slices = [30, 20, 25, 25]
labels = ['Math', 'Science', 'English', 'History']
colors = ['gold', 'lightblue', 'lightgreen', 'lightcoral']

plt.figure(figsize=(6, 6))
plt.pie(slices, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.axis('equal')  # Equal aspect ratio to ensure pie is a circle
plt.show()
```
