```python
# Plot Line plot with formatting to display two student's data with last 4 semester SPI. also use legend in plot, different line styles and marker characters.

import matplotlib.pyplot as plt

# Data
semesters = [1, 2, 3, 4]
student_A_spi = [7.8, 8.2, 8.5, 8.9]
student_B_spi = [8.0, 7.9, 8.3, 8.6]

# Plot for Student A
plt.plot(semesters, student_A_spi, 
         linestyle='--', marker='o', color='blue', 
         label='Student A')

# Plot for Student B
plt.plot(semesters, student_B_spi, 
         linestyle='-.', marker='s', color='green', 
         label='Student B')

# Labels and title
plt.xlabel("Semester")
plt.ylabel("SPI")
plt.title("SPI Progress Over Last 4 Semesters")

# Axis limits (optional)
plt.ylim(7.5, 9.5)
plt.xticks(semesters)

# Grid, Legend, and Display
plt.grid(True)
plt.legend()
plt.show()
```
