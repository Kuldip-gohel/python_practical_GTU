```python
# Draw a graph which satisfied following condition given
# practical.
# Give the limit to x-axis (-2,12)
# Give the limit to y-axis (-2,2)
# Display Annotation

import matplotlib.pyplot as plt
import math

# Data
x = list(range(10))
y = [math.cos(i) for i in x]

# Plot
plt.plot(x, y, linestyle='-.', marker='o')

# Axis limits
plt.xlim(-2, 12)
plt.ylim(-2, 2)

# Labels and title
plt.xlabel("this is X dimension")
plt.ylabel("this is Y dimension")
plt.title("This is My First Graph")

# Annotation: Annotate the first point
plt.annotate('Start Point',
             xy=(0,1),              # Point to annotate
             xytext=(3,1.5), # Location of the text
             arrowprops=dict(facecolor='black', arrowstyle='->'))

# Display grid and plot
plt.grid(True)
plt.show()
```
